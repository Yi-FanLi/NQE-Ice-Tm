# SET3-trained NEP density trajectories

Figure S1 design: instantaneous densities green, accumulated averages purple, same-condition DP reference means black dashed. Full trajectories are shown; averages start after200ps water/50ps ice. The two panels use different time ranges because trajectory lengths are2ns/500ps.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
ROOT=Path.cwd()
if not (ROOT/'energy_reference.json').exists(): ROOT=ROOT.parent
DATA=np.load(ROOT/'data/nep_set3/plot_data.npz')
plt.rcParams.update({'font.family':'Times New Roman','mathtext.fontset':'stix','font.size':12,'axes.labelsize':12,'axes.titlesize':12,'legend.fontsize':11,'xtick.labelsize':11,'ytick.labelsize':11,'pdf.fonttype':42,'ps.fonttype':42})
fig,axes=plt.subplots(2,1,figsize=(6.6,6.0),sharey=True)
for ax,phase,label,discard,dp,panel in zip(axes,['water','ice'],['liq','ice'],[200,50],[.900,.874],['(a)','(b)']):
    values=DATA[phase+'_density'];time=values[:,0];rho=values[:,1];keep=time>discard
    retained=rho[keep];avg=np.cumsum(retained)/np.arange(1,len(retained)+1)
    ax.plot(time,rho,color='#12a182',lw=.65,label=label+', NEP',rasterized=True)
    ax.plot(time[keep],avg,color='#525288',lw=1.8,label=label+', accumulated average')
    ax.axhline(dp,color='black',ls='--',lw=1.4,label=r'$\rho_{\mathrm{%s}}^{\mathrm{DP}}=%.3f$ g/cm$^3$'%(label,dp))
    ax.set_xlim(0,time[-1]);ax.set_ylim(.79,1.10)
    ax.set_ylabel(r'Density $\rho$ [g/cm$^3$]');ax.set_xlabel(r'Time $t$ [ps]')
    ax.text(.012,.96,panel,transform=ax.transAxes,va='top')
    ax.legend(loc='upper right',frameon=False,fontsize=10)
    ax.tick_params(direction='in',top=True,right=True)
fig.subplots_adjust(hspace=.29,left=.13,right=.98,bottom=.1,top=.98)
fig.savefig(ROOT/'figures/nep_set3_density_trajectories.pdf',bbox_inches='tight')
fig.savefig(ROOT/'figures/nep_set3_density_trajectories.png',dpi=600,bbox_inches='tight')
```
